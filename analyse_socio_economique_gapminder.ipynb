```python
import plotly.express as px
import pandas as pd
import plotly.io as pio
pio.renderers.default = "colab"
# 1. Chargement des données Dataset réel Gapminder
df = px.data.gapminder().query("year == 2007")

# GRAPHIQUE 1 : Scatter Plot avec Symboles (Santé vs Richesse)
fig1 = px.scatter(df, x="gdpPercap", y="lifeExp",
                 symbol="continent", color="continent",
                 hover_name="country", log_x=True,
                 title="Lien entre PIB/habitant et Espérance de vie (2007)",
                 labels={"gdpPercap": "PIB par habitant (Log)", "lifeExp": "Espérance de vie"})
fig1.show()

# GRAPHIQUE 2 : Bubble Chart (La dimension Population)
fig2 = px.scatter(df, x="gdpPercap", y="lifeExp",
                 size="pop", color="continent",
                 hover_name="country", size_max=60, log_x=True,
                 title="Bubble Chart : Santé, Richesse et Population",
                 labels={"pop": "Population Totale"})
fig2.show()

# GRAPHIQUE 3 : Carte Choroplèthe (Géographie de la richesse)
fig3 = px.choropleth(df, locations="iso_alpha",
                    color="gdpPercap",
                    hover_name="country",
                    color_continuous_scale=px.colors.sequential.Plasma,
                    title="Carte du Monde : Répartition du PIB par habitant")
fig3.show()
```

JUSTIFICATIONS DES CHOIX DES GRAPHIQUES

1. Le Scatter Plot avec Symboles (Le décodeur régional)
L'idée : On voulait voir si l'argent fait le bonheur (ou au moins une longue vie).
Ce qu'on voit : En utilisant des symboles différents (cercles, croix, carrés) pour chaque continent, on remarque que l'Europe est ultra-groupée en haut à droite. C'est une "bulle" de privilège. L'Afrique, représentée par un autre symbole, est plus dispersée mais majoritairement en bas à gauche. Ça montre que la géographie est un facteur énorme dans le développement.

2. Le Bubble Chart (L'effet de masse)
L'idée : Un point c'est bien, mais une bulle qui montre la population, c'est mieux.
Ce qu'on voit : Les énormes bulles de la Chine et de l'Inde au milieu du graphique changent toute notre perspective. On réalise que même si elles n'ont pas encore le PIB des USA, elles représentent une part massive de l'humanité qui progresse très vite. C'est le graphique le plus puissant pour comprendre les enjeux globaux.

3. La Carte Choroplèthe (La vue du ciel)
L'idée : On voulait voir où se trouve la richesse physiquement sur la planète.
Ce qu'on voit : C'est brutal. Le contraste de couleurs entre le Nord (très sombre/intense) et le Sud (plus clair) saute aux yeux. La carte permet de voir instantanément des zones de pauvreté ou de richesse régionale que les autres graphiques cachent.
